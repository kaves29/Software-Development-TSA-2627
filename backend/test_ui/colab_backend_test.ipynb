```python
import os
import subprocess
from pathlib import Path

project = Path('/content/speakly-backend-test')
if project.exists():
    raise RuntimeError('The Speakly folder already exists in this Colab session. Restart the session before running again.')
subprocess.run(['git', 'clone', '--depth', '1', '--branch', 'main',
                'https://github.com/kaves29/Software-Development-TSA-2627.git', str(project)],
               check=True)
os.chdir(project)
subprocess.run(['python', '-m', 'pip', 'install', '-q', '-r', 'backend/requirements.txt'], check=True)
print('Speakly is installed for this session.')
```

```python
import subprocess
import sys
import time
import urllib.request

from google.colab import output

server = subprocess.Popen([sys.executable, '-m', 'uvicorn', 'backend.main:app',
                           '--host', '127.0.0.1', '--port', '8765'])
for attempt in range(60):
    if server.poll() is not None:
        raise RuntimeError('The server stopped before it was ready. Check the output above.')
    try:
        with urllib.request.urlopen('http://127.0.0.1:8765/health', timeout=2) as response:
            if response.status == 200:
                break
    except (OSError, TimeoutError):
        time.sleep(1)
else:
    server.terminate()
    raise RuntimeError('The server did not start in time. Check the output above.')

print('The server is ready. The test page is below.')
output.serve_kernel_port_as_iframe(8765, path='/test', height=1000)
```
